## The learnable signal: the correct answer appears in the support far more than distractors

```python
def frac_in(col):
    return np.mean([str(a).lower() in str(s).lower() for a, s in zip(sciq[col], sciq['support'])])
{c: round(frac_in(c), 3) for c in ['correct_answer','distractor1','distractor2','distractor3']}
```

## AI Flashcards data-quality note
Why we do NOT train classifiers on this dataset: `difficulty` is statistically independent of the other labels (uniform crosstab), and `subject`/`topic` are written verbatim into `source_text`.

```python
aif = load_raw_ai_flashcards(); en = aif[aif['language'] == 'en']
pd.crosstab(en['difficulty'], en['bloom_level'], normalize='index').round(3)
```
